```python
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt, pi, exp

# Input and data generation
mean = float(input("Enter mean value: "))
std_dev = float(input("Enter standard deviation: "))
data = np.random.normal(mean, std_dev, 1_000_000)
sample_std = np.std(data)

# Calculate Gaussian probabilities
start_point = mean - 4 * std_dev
end_point = mean + 4 * std_dev
x_values = np.linspace(start_point, end_point, 100)
probabilities = [(1 / (std_dev * sqrt(2 * pi))) * exp(-0.5 * ((x - mean) / std_dev) ** 2) for x in x_values]

# Print statistics
print(f"\nMean: {mean}")
print(f"Standard deviation (input): {std_dev:.4f}")
print(f"Sample standard deviation: {sample_std:.4f}")
print("-" * 50)

# Visualization
plt.figure(figsize=(12, 6))
plt.plot(x_values, probabilities, color='navy', linewidth=2)

plt.title(f'Gaussian Distribution (μ = {mean}, σ = {std_dev})', fontsize=14)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add theoretical vs sample std deviation annotation
plt.annotate(f'Input Std: {std_dev:.2f}\nSample Std: {sample_std:.2f}',
             xy=(0.7, 0.9), xycoords='axes fraction',
             bbox=dict(boxstyle='round', fc='white'))

# Fill area under curve
plt.fill_between(x_values, probabilities, color='skyblue', alpha=0.4)

plt.tight_layout()
plt.savefig('gaussian_distribution.png', dpi=300)
plt.show()
```
